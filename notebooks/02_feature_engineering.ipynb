```python
import pandas as pd 
```

```python
df = pd.read_csv("../data/processed/cleaned_data.csv", )
```

Encoding. 

```python
df = df.drop(columns='Unnamed: 0')
```

```python
df.select_dtypes(exclude='number').nunique()
```

```python
df.info(verbose=True, show_counts=True)
```

Start by looking at higher cardinality features.

```python
for col in ['City', 'Zip_Code', 'Offer', 'Payment_Method', 'Internet_Type']: 
    counts = df[col].value_counts()
    print(col, '\n| unique: ', len(counts), 
          '\n| median rows per value', counts.median(), 
          '\n| percent of values less than five percent of customers', round((counts < round((len(df) / 100) * 5)).mean() * 100, 1))
```

```python
df = df.drop(columns = ['City', 'Zip_Code', 'Segment'])
```

```python
df = df.assign(Is_Male = df['Gender'].map({'Male': True, 'Female': False})).drop(columns = 'Gender')
```

```python
df.isna().sum().sum()
```

```python
df.head().T
```

Okay I now want to encode boolean values in (0/1).

```python
bool_cols = df.select_dtypes(include = 'bool').columns 
```

```python
df[bool_cols] = df[bool_cols].apply(lambda x : x.astype(float))
```

```python
df.info()
```

```python
df.select_dtypes(include='str').nunique()
```

```python
df['Contract'].unique()
```

```python
df['Contract'] = df['Contract'].map({'Month-to-Month': 0, 'One Year': 1, 'Two Year': 2}).astype(int)
```
